```python
#fixing datetimes [ return the value as date : time ]
df["Date_of_Admission"]=pd.to_datetime(df["Date_of_Admission"])
df["Discharge_Date"]=pd.to_datetime(df["Discharge_Date"])

```

```python
#duration of stay

df["Days_Stayed"]=(df["Discharge_Date"]-df["Date_of_Admission"]).dt.days
df["Days_Stayed"].describe()
```

```python
#remove the time from the baove changed date time format because time shows 00:00:00 values 
df["Date_of_Admission"]=df["Date_of_Admission"].dt.date
df["Discharge_Date"]=df["Discharge_Date"].dt.date
```

```python
#text_Cleaning
text_columns=["Name","Gender","Blood_Type","Medical_Condition","Doctor","Hospital","Insurance_Provider","Admission_Type","Medication","Test_Results"]

for col in text_columns:
    df[col]=df[col].astype(str).str.strip()
```

```python
age_bins=[0,12,19,59,100] 
# inga 5 values en kuduthurkangaa naa 2 values ku edaila irkura range like 0-12 => child 
# 12-19 => teenager , 19-59 => adults and 59-100 => seniors 

labels=['Child','Teenage','Adult','Senior']

df["Age_Group"]=pd.cut(df["Age"],bins=age_bins, labels=labels)

print(df["Age_Group"].value_counts())
```

```python
df["Age_Group"]
```

```python
#fixing datetimes [ return the value as date : time ]
df["Date_of_Admission"]=pd.to_datetime(df["Date_of_Admission"])
df["Discharge_Date"]=pd.to_datetime(df["Discharge_Date"])


df["Date_of_Admission"]=df["Date_of_Admission"].dt.date
df["Discharge_Date"]=df["Discharge_Date"].dt.date
```

```python
df.to_excel("healthcare data cleaned.xlsx", index=False)
print("Cleaned Data saved Successfully")
```

```python
df_clean=pd.read_excel("healthcare data cleaned.xlsx")

df_clean.head()
```

```python
import matplotlib.pyplot as plt

import seaborn as sns 

sns.countplot(x="Age_Group", data=df_clean)
plt.title("Patient count by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of patients")
plt.show()
```

```python
sns.barplot(x="Medical_Condition", y="Billing_Amount", data=df_clean, errorbar=None)
plt.title("Average Billing amount by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Average Billing Amount ($)")
plt.show()
```

```python
sns.barplot(x="Days_Stayed", y="Medical_Condition", data=df_clean,errorbar=None)
plt.title("Patients Duration of Stay based on Medical COndition")
plt.xlabel("Days Stayed")
plt.ylabel("Medical Condition")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
